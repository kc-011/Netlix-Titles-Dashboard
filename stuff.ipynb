```python
from datetime import timedelta
weather = pd.DataFrame({'id':[1,2,3,4],'recordDate':['2015-1-31','2015-2-1','2015-2-3','2015-2-4'],'temperature':[10,25,20,30]})
true_id = []
weather['recordDate'] = pd.to_datetime(weather['recordDate'])
weather = weather.sort_values(by='recordDate',ignore_index=True)

for i in range(1,len(weather['id'])):
    if weather['recordDate'][i] - weather['recordDate'][i-1] == timedelta(1):
        if weather['temperature'][i] > weather['temperature'][i-1]:
            true_id.append(weather['id'][i])

pd.DataFrame({'id':true_id})
# weather['recordDate'][1].day - weather['recordDate'][0].day == 1
```

```python
s = "()(())((()()))"
j = 0
lst = []
for i in s:
    if i == '(':
        j+=1
        lst.append(j)
    elif i == ')':
        j-=1
        lst.append(j)
    else:
        continue

print(max(lst) if lst else 0)
print(lst)
```
